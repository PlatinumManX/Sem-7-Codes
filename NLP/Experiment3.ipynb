```python
import re
from bs4 import BeautifulSoup

text = """
<h1>Hello World!</h1>
This is   an example of NLP!!!   preprocessing.
It removes <b>HTML tags</b> and @#$ special characters.
"""

print("Original Text:")
print(text)

# Remove HTML tags
text = BeautifulSoup(text, "html.parser").get_text()

# Remove special characters
text = re.sub(r'[^A-Za-z0-9\s]', '', text)

# Remove extra whitespaces
text = re.sub(r'\s+', ' ', text).strip()

print("\nPreprocessed Text:")
print(text)
```

```python
text = """
I am learning NLP. I can't understand everything at first.
NLP is interesting! It won't be difficult if I practice regularly.
"""
print("original sentence")
print(text)
# Expand contractions
contractions = {
    "can't": "cannot",
    "won't": "will not",
    "I'm": "I am",
    "don't": "do not",
    "isn't": "is not",
    "it's": "it is"
}

for contraction, expansion in contractions.items():
    text = text.replace(contraction, expansion)

# Sentence boundary detection
sentences = nltk.sent_tokenize(text)

print("Sentences:")
for i, sentence in enumerate(sentences, 1):
    print(i, ":", sentence)
```

```python
import nltk
import re

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

text = "I have Twenty students and 5 books. They are studying NLP in the classroom."

print("Original Text:")
print(text)

# 1. Convert text to lowercase
text = text.lower()

# 2. Convert number words to numeric form
number_words = {
    "one": "1",
    "two": "2",
    "three": "3",
    "four": "4",
    "five": "5",
    "six": "6",
    "seven": "7",
    "eight": "8",
    "nine": "9",
    "ten": "10",
    "twenty": "20"
}

for word, number in number_words.items():
    text = re.sub(r'\b' + word + r'\b', number, text)

print("\nAfter Number Conversion:")
print(text)

# 3. Remove numbers
text = re.sub(r'\d+', '', text)
print("\nafter  number removal:")
print(text)
# 4. Tokenization
tokens = word_tokenize(text)
print("After tokenisation") 
print(tokens)
# 5. Remove stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [
    word for word in tokens
    if word.lower() not in stop_words and word.isalpha()
]

print("\nFinal Text After Stopword:")
print(filtered_words)
```

```python
from nltk.tokenize import word_tokenize
from nltk import pos_tag

text = "Natural language processing is an interesting field of computer science."

# 1. Tokenization
tokens = word_tokenize(text)

print("Tokens:")
print(tokens)

# 2. POS Tagging
pos_tags = pos_tag(tokens)

print("\nPOS Tags:")
print(pos_tags)

# 3. Phrase Extraction
# Extract noun phrases using POS tags
noun_phrases = []

for i in range(len(pos_tags) - 1):
    word1, tag1 = pos_tags[i]
    word2, tag2 = pos_tags[i + 1]

    if tag1 in ['JJ', 'NN', 'NNS'] and tag2 in ['NN', 'NNS']:
        phrase = word1 + " " + word2
        noun_phrases.append(phrase)

print("\nExtracted Phrases:")
for phrase in noun_phrases:
    print(phrase)

# 4. Script Validation
print("\nScript Validation:")

for word in tokens:
    if re.fullmatch(r'[A-Za-z]+', word):
        print(word, "-> English Script")
    else:
        print(word, "-> Non-English/Special Character")
```

```python
import re
import matplotlib.pyplot as plt

text = """
Hello!!! This is an NLP example.
I have 25 students, and they are learning NLP.
Visit https://example.com for more information.
"""

print("Original Text:")
print(text)

# 1. Remove URLs
clean_text = re.sub(r'https?://\S+|www\.\S+', '', text)

# 2. Remove special characters
clean_text = re.sub(r'[^A-Za-z0-9\s]', '', clean_text)

# 3. Convert to lowercase
clean_text = clean_text.lower()

# 4. Remove numbers
clean_text = re.sub(r'\d+', '', clean_text)

# 5. Remove extra spaces
clean_text = re.sub(r'\s+', ' ', clean_text).strip()

print("\nPreprocessed Text:")
print(clean_text)

# Count words before and after preprocessing
original_words = len(text.split())
processed_words = len(clean_text.split())

print("\nOriginal Word Count:", original_words)
print("Processed Word Count:", processed_words)

# Visualization
labels = ['Before Preprocessing', 'After Preprocessing']
values = [original_words, processed_words]

plt.bar(labels, values)
plt.xlabel("Text")
plt.ylabel("Number of Words")
plt.title("Impact of Text Preprocessing")
plt.show()
```
