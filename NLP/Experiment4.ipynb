```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

text = "The students are studying and learning different programming concepts."

# Tokenization
words = word_tokenize(text.lower())

# Stopword removal
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word not in stop_words]

# Stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]

# Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print("Original Text:")
print(text)

print("\nAfter Stopword Removal:")
print(filtered_words)

print("\nAfter Stemming:")
print(stemmed_words)

print("\nAfter Lemmatization:")
print(lemmatized_words)
```

```python
import re

text = """
Hello!!! My name is Rahul Sharma.
My email is rahul@gmail.com and my phone number is 9876543210.
I am learning NLP. नमस्ते, मैं NLP सीख रहा हूँ।
"""

print("Original Text:")
print(text)

# 1. Noise removal
clean_text = re.sub(r'[!@#$%^&*(),.?":{}|<>]', '', text)
clean_text = re.sub(r'\s+', ' ', clean_text).strip()

print("\nAfter Noise Removal:")
print(clean_text)

# 2. Multilingual text processing
print("\nMultilingual Text:")
for word in clean_text.split():
    if re.search(r'[\u0900-\u097F]', word):
        print(word, "-> Hindi")
    elif re.search(r'[A-Za-z]', word):
        print(word, "-> English")

# 3. Entity anonymization
# Email anonymization
anonymized_text = re.sub(
    r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b',
    '[EMAIL]',
    clean_text
)

# Phone number anonymization
anonymized_text = re.sub(
    r'\b\d{10}\b',
    '[PHONE]',
    anonymized_text
)

# Name anonymization
anonymized_text = anonymized_text.replace(
    "Rahul Sharma",
    "[PERSON]"
)

print("\nAfter Entity Anonymization:")
print(anonymized_text)
```
